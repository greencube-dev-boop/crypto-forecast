# 🚀 Multi-Horizon Cryptocurrency Price Forecasting System & Colab Web GUI

Predicts cryptocurrency prices across **11 discrete horizons**:
`1h`, `2h`, `4h`, `6h`, `12h`, `24h`, `2d`, `3d`, `7d`, `14d`, `30d`.

### Supported Engines & Capabilities:
- **Google TimesFM 2.5 (200M)**: Zero-shot time-series foundation transformer
- **Amazon Chronos-2 & Chronos-Bolt**: Probabilistic quantile trajectories (P10, P50, P90)
- **IBM Granite TTM**: Tiny Time Mixers for low-latency multi-coin screening
- **ElKulako/CryptoBERT**: News & social sentiment drift conviction attenuation
- **Interactive Web GUI**: Plotly candlestick & uncertainty cones with public `share=True` Gradio tunneling
